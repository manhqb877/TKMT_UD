### Bài 1: Tạo quần thể mô phỏng

Bạn hãy tạo một mẫu gồm 10000 phần tử mô phỏng chiều cao của nam thanh niên có giá trị từ 120cm - 200cm. Bạn hãy lưu kết quả vào biến **POP**.

Tính:
a. Chiều cao trung bình (kỳ vọng) của quần thể và độ lệch chuẩn về chiều cao của quần thể.

b. Tính tỷ lệ người cao trong quần thể, biết rằng thanh niên có chiều cao từ 180 trở lên được gọi là cao.

c. Vẽ histogram về chiều cao của quần thể. Theo bạn quần thể có phân phối chuẩn hay không?

d. Thử vẽ histogram và đồ thị hàm mật độ của phân phối chuẩn sử dụng tham số loc và scale bằng với kỳ vọng và độ lệch chuẩn của quần thể.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
POP = np.random.uniform(120, 200, 10000)
#a
mean_height = np.mean(POP)
std_height = np.std(POP)

print(f"Chiều cao trung bình: {mean_height:.2f} cm")
print(f"Độ lệch chuẩn: {std_height:.2f} cm")
#b
high_people_ratio = np.sum(POP >= 180) / len(POP)
print(f"Tỷ lệ người cao trong quần thể: {high_people_ratio:.2%}")
#c
plt.figure(figsize=(10, 6))
sns.histplot(POP, kde=True, bins=50, color='blue')
plt.title('Histogram về chiều cao của quần thể')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Tần suất')
plt.show()
#d
x = np.linspace(120, 200, 100)
normal_dist = (1/(std_height * np.sqrt(2 * np.pi))) * np.exp(-(x - mean_height)**2 / (2 * std_height**2))

plt.figure(figsize=(10, 6))
sns.histplot(POP, kde=True, bins=50, color='blue', stat='density', label='Histogram')
plt.plot(x, normal_dist, label=f'Phân phối chuẩn\n(loc={mean_height:.2f}, scale={std_height:.2f})', color='red', linewidth=2)
plt.title('So sánh histogram với phân phối chuẩn')
plt.xlabel('Chiều cao (cm)')
plt.ylabel('Mật độ')
plt.legend()
plt.show()
```

### Bài 2: Lấy mẫu cỡ 20

b. Tỷ lệ người cao của mẫu

c. Vẽ histogram của mẫu

### Bài 4: Khảo sát phân phối của trung bình mẫu

a. Thực hiện bài 2 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

b. Thực hiện bài 3 100 lần, mỗi lần bạn tính được trung bình mẫu. Vẽ đồ thị histogram cho 100 trung bình mẫu bạn tính được.

c. Vẽ histogram của hai câu a, b trong cùng một hình để so sánh. Theo bạn hình dạng của histogram thay đổi nói lên điều gì?

Gợi ý: tạo một danh sách rỗng để chứa các trung bình mẫu tính được, sử dụng hàm append() để thêm giá trị trung bình vào danh sách sau mỗi lần tính.

```python

# Bước 1: Tạo quần thể POP với 10000 phần tử, chiều cao từ 120cm đến 200cm
POP = np.random.uniform(120, 200, 10000)

# a. Thực hiện bài 2 100 lần (cỡ mẫu 20), mỗi lần tính được trung bình mẫu
sample_means_20 = []  # Danh sách để chứa trung bình mẫu của cỡ mẫu 20
for _ in range(100):
    sample_20 = np.random.choice(POP, size=20)
    sample_means_20.append(np.mean(sample_20))

# b. Thực hiện bài 3 100 lần (cỡ mẫu 100), mỗi lần tính được trung bình mẫu
sample_means_100 = []  # Danh sách để chứa trung bình mẫu của cỡ mẫu 100
for _ in range(100):
    sample_100 = np.random.choice(POP, size=100)
    sample_means_100.append(np.mean(sample_100))

# c. Vẽ histogram của hai câu a, b trong cùng một hình để so sánh
plt.figure(figsize=(12, 6))

# Vẽ histogram cho trung bình mẫu của cỡ mẫu 20
sns.histplot(sample_means_20, kde=True, bins=20, color='skyblue', label='Cỡ mẫu 20', stat='density')

# Vẽ histogram cho trung bình mẫu của cỡ mẫu 100
sns.histplot(sample_means_100, kde=True, bins=20, color='orange', label='Cỡ mẫu 100', stat='density')

# Thêm tiêu đề và nhãn
plt.title('So sánh phân phối của trung bình mẫu (Cỡ mẫu 20 vs 100)')
plt.xlabel('Trung bình mẫu')
plt.ylabel('Mật độ')
plt.legend()

# Hiển thị đồ thị
plt.show()
```

### Bài 6: Khảo sát phương sai mẫu

Bạn hãy mô phỏng 1000 lần lấy mẫu có kích cỡ 100. Vẽ histogram đại lượng phương sai của các mẫu (Phương sai bằng bình phương độ lệch chuẩn)

```python

# Bước 1: Tạo quần thể POP với 10000 phần tử, chiều cao từ 120cm đến 200cm
POP = np.random.uniform(120, 200, 10000)

# Bước 2: Mô phỏng 1000 lần lấy mẫu cỡ 100 và tính phương sai của các mẫu
variances_100 = []  # Danh sách để chứa phương sai của các mẫu cỡ 100
for _ in range(1000):
    sample_100 = np.random.choice(POP, size=100)
    sample_variance = np.var(sample_100)  # Tính phương sai của mẫu
    variances_100.append(sample_variance)

# Bước 3: Vẽ histogram của phương sai của các mẫu
plt.figure(figsize=(10, 6))

# Vẽ histogram cho phương sai của các mẫu cỡ 100
sns.histplot(variances_100, kde=True, bins=30, color='skyblue', stat='density')

# Thêm tiêu đề và nhãn
plt.title('Phân phối phương sai của các mẫu (Cỡ mẫu 100)')
plt.xlabel('Phương sai')
plt.ylabel('Mật độ')

# Hiển thị đồ thị
plt.show()
```

### Bài 7:  Bài Tập Ước Lượng 

Khảo sát cân nặng (kg) của gà khi xuất chuồng, người ta cân một số con và có kết quả như sau:

> 2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8

Giả sử cân nặng của gà là biến ngẫu nhiên có phân phối chuẩn. Hãy lập khoảng tin cậy cho cân nặng trung bình của  khi xuất chuồng với độ tin cậy là 95%. Trong 2 trường hợp:

a, Biết $\sigma = 0.3$

b, Không biết $\sigma$


```python
import numpy as np
import scipy.stats as stats

# Dữ liệu cân nặng của gà
weights = np.array([2.1, 1.8, 2.0, 2.3, 1.7, 1.5, 2.0, 2.2, 1.8])

# Kích thước mẫu (n)
n = len(weights)

# Trung bình mẫu (x_bar)
x_bar = np.mean(weights)

# Độ lệch chuẩn mẫu (s)
s = np.std(weights, ddof=1)

# a. Biết sigma = 0.3, khoảng tin cậy cho trung bình (95% tin cậy)
sigma = 0.3
z_alpha_2 = stats.norm.ppf(0.975)  # Z-value cho mức độ tin cậy 95%
margin_of_error_a = z_alpha_2 * (sigma / np.sqrt(n))
ci_a = (x_bar - margin_of_error_a, x_bar + margin_of_error_a)

print(f"(a) Khoảng tin cậy với sigma = 0.3: {ci_a}")

# b. Không biết sigma, khoảng tin cậy cho trung bình (95% tin cậy)
t_alpha_2 = stats.t.ppf(0.975, df=n-1)  # t-value cho mức độ tin cậy 95% và df = n-1
margin_of_error_b = t_alpha_2 * (s / np.sqrt(n))
ci_b = (x_bar - margin_of_error_b, x_bar + margin_of_error_b)

print(f"(b) Khoảng tin cậy với không biết sigma: {ci_b}")
```

### Bài 8

Bộ dữ liệu `SAT-lard.csv` chứa điểm thi SAT của 1000 học sinh. 

a. Giả sử bộ dữ liệu là kết quả của quá trình lấy mẫu ngẫu nhiên từ các học sinh. Bạn hãy xây dựng khoảng tin cậy cho trung bình điểm thi của tất cả các sinh viên với độ tin cậy 99% (`Đáp án: 105.2 ± 3.10`)

b. Giả sử bộ dữ liệu là kết quả của quá trình lấy mẫu ngẫu nhiên từ các học sinh. Bạn hãy xây dựng khoảng tin cậy cho trung bình điểm thi của tất cả các sinh viên với độ tin cậy (`Đáp án: 105.2 ± 1.86`)

c. Giả sử bộ dữ liệu là toàn bộ kết quả điểm thi SAT của học sinh một trường phố thông. Tính điểm SAT trung bình của trường phổ thông đấy. (`Đáp án:  μ = 1528.74`)

d. Giả sử bộ dữ liệu là toàn bộ kết quả điểm thi SAT của học sinh một trường phố thông. Sử dụng mẫu là điểm thi của 36 học sinh đầu tiên trong bộ dữ liệu, bạn hãy xây dựng khoảng tin cậy **99%** cho điểm SAT trung bình của trường phổ thông đấy. (`Đáp án: (1428.22, 1602.89)`)

e. Giả sử bộ dữ liệu là toàn bộ kết quả điểm thi SAT của học sinh một trường phố thông. Sử dụng mẫu là điểm thi của 36 học sinh đầu tiên trong bộ dữ liệu, bạn hãy xây dựng khoảng tin cậy **95%** cho điểm SAT trung bình của trường phổ thông đấy. Cho biết khoảng ước lượng bạn xây dựng có đúng không, dựa trên giá trị đã tính ở câu trên.

```python
import numpy as np
import pandas as pd
import scipy.stats as stats

df = pd.read_csv("SAT-lard.csv")

print(df.head())

sat_scores = df["SAT Score"]

# a. Khoảng tin cậy cho trung bình điểm thi với độ tin cậy 99%
n = len(sat_scores)  # Số lượng học sinh
x_bar = np.mean(sat_scores)  # Trung bình mẫu
s = np.std(sat_scores, ddof=1)  # Độ lệch chuẩn mẫu

# Độ tin cậy 99% => Z-value cho mức tin cậy 99%
z_alpha_2 = stats.norm.ppf(0.995)  # Z-value cho mức độ tin cậy 99%

# Sai số giới hạn và khoảng tin cậy
margin_of_error_a = z_alpha_2 * (s / np.sqrt(n))
ci_a = (x_bar - margin_of_error_a, x_bar + margin_of_error_a)

print(f"(a) Khoảng tin cậy 99%: {ci_a}")

# b. Khoảng tin cậy cho trung bình điểm thi với độ tin cậy 95%
z_alpha_2_95 = stats.norm.ppf(0.975)  # Z-value cho mức độ tin cậy 95%

# Sai số giới hạn và khoảng tin cậy
margin_of_error_b = z_alpha_2_95 * (s / np.sqrt(n))
ci_b = (x_bar - margin_of_error_b, x_bar + margin_of_error_b)

print(f"(b) Khoảng tin cậy 95%: {ci_b}")

# c. Tính điểm SAT trung bình của trường (nếu đây là toàn bộ dữ liệu)
mu = x_bar  # Vì đây là toàn bộ dữ liệu, trung bình mẫu chính là trung bình của toàn bộ quần thể
print(f"(c) Điểm SAT trung bình của trường: {mu}")

# d. Khoảng tin cậy 99% cho điểm SAT trung bình của trường dựa trên mẫu 36 học sinh đầu tiên
sample_36 = sat_scores[:36]  # Dùng mẫu 36 học sinh đầu tiên

# Các tham số của mẫu 36 học sinh
n_36 = len(sample_36)
x_bar_36 = np.mean(sample_36)
s_36 = np.std(sample_36, ddof=1)

# Độ tin cậy 99% => t-value cho mức tin cậy 99% và bậc tự do df = n-1
t_alpha_2_36 = stats.t.ppf(0.995, df=n_36 - 1)

# Sai số giới hạn và khoảng tin cậy
margin_of_error_d = t_alpha_2_36 * (s_36 / np.sqrt(n_36))
ci_d = (x_bar_36 - margin_of_error_d, x_bar_36 + margin_of_error_d)

print(f"(d) Khoảng tin cậy 99% cho mẫu 36 học sinh đầu tiên: {ci_d}")

# e. Khoảng tin cậy 95% cho mẫu 36 học sinh đầu tiên và so sánh
t_alpha_2_36_95 = stats.t.ppf(0.975, df=n_36 - 1)

# Sai số giới hạn và khoảng tin cậy
margin_of_error_e = t_alpha_2_36_95 * (s_36 / np.sqrt(n_36))
ci_e = (x_bar_36 - margin_of_error_e, x_bar_36 + margin_of_error_e)

print(f"(e) Khoảng tin cậy 95% cho mẫu 36 học sinh đầu tiên: {ci_e}")
```
