b. (2 điểm) Mỗi câu trả lời đúng được 0.25 điểm. Tính xác suất thí sinh đạt được từ 4 điểm trở lên.

```python
# a
import numpy as np
from scipy import stats
n, tle = 40,  1/4
p1 = 20
p2 = 16
print(f"(a)Xác suất thí sinh trả lời đúng 20 câu hỏi :  {stats.binom.pmf(p1, n, tle):.4f}")
print(f"(b)xác suất thí sinh đạt được từ 4 điểm trở lên :  {stats.binom.cdf(p2-1,n,tle):.4f}")
```

Câu 2. Để đánh giá kết quả học tập cuối năm của học sinh lớp 12 của tỉnh A, người ta chọn ngẫu nhiên 40 học sinh, điểm tổng kết của các học sinh như sau:
$5.1, 5.5, 7.6, 8.0, 8.5, 6.5, 6.6, 7.2, 4.8, 7.5, 8.8, 5.7, 6.9, 8.2, 7.7, 6.3, 5.6, 7.0, 7.4, 7.3, 6.9, 5.7, 5.3, 6.0, 5.0, 8.4, 9.2, 7.4, 7.9, 5.9, 
6.6, 7.8, 8.7, 8.8, 6.8, 7.1, 7.5, 6.8, 5.8, 6.4$

a. (3 điểm) Tìm khoảng ước lượng điểm tổng kết trung bình của học sinh lớp 12 của tỉnh A với độ tin cậy 90%, 95%, 99%.

b. (1 điểm) Tìm khoảng ước lượng phương sai điểm tổng kết của học sinh lớp 12 của tỉnh A với độ tin cậy 95%.

c. (1 điểm) Những học sinh có điểm trung bình từ 8.0 trở lên là học sinh giỏi. Hãy ước lượng tỷ lệ học sinh giỏi của tỉnh A với độ tin cậy 99%.

d. (2 điểm) Một báo cáo công bố rằng, tỷ lệ học sinh giỏi lớp 12 của tỉnh A là 27%. Với mức ỹ nghĩa 1%, báo cáo trên có đáng tin không?

```python
# Câu a
print("\nCâu a:")
import numpy as np
from scipy import stats
radius_tong = np.array([5.1, 5.5, 7.6, 8.0, 8.5, 6.5, 6.6, 7.2, 4.8, 7.5, 8.8, 5.7, 6.9, 8.2, 7.7, 6.3, 5.6, 7.0, 7.4, 7.3, 6.9, 5.7, 5.3, 6.0, 5.0, 8.4, 9.2, 7.4, 7.9, 5.9, 
6.6, 7.8, 8.7, 8.8, 6.8, 7.1, 7.5, 6.8, 5.8, 6.4])

n = len(radius_tong)
tb = np.mean(radius_tong)
doLechChuan = np.std(radius_tong, ddof=1)
dtc = [0.90, 0.95, 0.99]

for tc in dtc:
    al = 1 - tc
    t_cr = stats.t.ppf(1-al/2, df = n - 1)
    mar = t_cr * (doLechChuan/np.sqrt(n))
    lo = tb - mar
    up_b = tb + mar
    print(f'Độ tin Cậy {int(tc * 100)}%:({lo:.4f}, {up_b:.4f})')


# check = np.sum(radius_tong)
# n = len(radius_tong)
# print(f'Tong : {check:.4f}')
# print(f'Uoc luong trung binh voi do tin cay la : 90%')
# z_90 = stats.norm.ppf(0.90)  
# margin_error = z_90 * (check / np.sqrt(n))
# print(f"Giá trung bình: {check:.2f} ± {margin_error:.2f}")

# print(f'Uoc luong trung binh voi do tin cay la : 95%')
# z_95 = stats.norm.ppf(0.95)  
# margin_error = z_95 * (check / np.sqrt(n))
# print(f"Giá trung bình: {check:.2f} ± {margin_error:.2f}")

# print(f'Uoc luong trung binh voi do tin cay la : 99%')
# z_99 = stats.norm.ppf(0.99)  
# margin_error = z_99 * (check / np.sqrt(n))
# print(f"Giá trung bình: {check:.2f} ± {margin_error:.2f}")

# Câu c :

print("\nCâu c:")

Excellent = np.sum(radius_tong >= 8)
tlmau = Excellent / n
als = 0.01
z_f = stats.norm.ppf(1- als/2)
mar_f = z_f * np.sqrt(tlmau * (1-tlmau) /n)
lz = tlmau - mar_f
ipz = tlmau + mar_f
print(f'tỷ lệ học sinh giỏi của tỉnh A với độ tin cậy 99% :({lz:.4f}, {ipz:.4f})')
# Câu d :
print(f'Câu d')
# Giả thuyết
# H(0) : Tỷ lệ học ính giỏi của tỉnh A là 27% ( có nghĩa là u = u0 = 0.27)
# H(1) : Tỷ lệ học ính giỏi của tỉnh A khác 27% ( u != 0.27)


# Thông số
n = len(radius_tong)  
x_bar = np.mean(radius_tong)  # Trung bình mẫu
sigma = np.std(radius_tong, ddof=1)  # Độ lệch chuẩn mẫu

mu_0 = 0.27
alpha = 0.01  # Mức ý nghĩa

# Tính giá trị z
standard_error = sigma / np.sqrt(n)
z_stat = (tlmau - mu_0) / np.sqrt(mu_0 * (1-mu_0) / n)
# z_critical = 2 * (1-stats.norm.cdf(abs(z_stat)))
z_critical1 =  stats.norm.cdf(1-alpha /2)
print(f"Giá trị z: {z_stat:.4f}")

# z_critical = stats.norm.ppf(1 - alpha / 2)
# print(f"Giá trị tới hạn: ±{z_critical:.4f}")
print(f"Giá trị tới hạn: ±{z_critical1:.4f}")

if abs(t_stat) > z_critical1:
    print("H0: Với mức ỹ nghĩa 1%, báo cáo trên có đáng tin là tỷ  lệ học ính giỏi của tỉnh A là 27%")
else:
    print("H1: Với mức ỹ nghĩa 1%, báo cáo trên không  đáng tin là tỷ Tỷ lệ học ính giỏi của tỉnh A khác 27%")
```
