Bạn hãy phát sinh ngẫu nhiên mảng ndarray có tên `arr` gồm 100 phần tử số nguyên có giá trị từ 0..1000. Thực hiện các thao tác sau:

### 1. Hiển thị các phần tử thỏa yêu cầu dưới đây

- 10 phần tử đầu tiên
- 5 phần tử cuối cùng
- Các phần tử thỏa mãn điều kiện chia hết cho 5 và lớn hơn 100

```python
import numpy as np
from numpy import *

# Tạo mảng gồm 100 số nguyên từ 0..1000
x = random.randint(1000, size=100)
y = x
# Các phần tử thỏa mãn điều kiện chia hết cho 5 và lớn hơn 100
for i in y:
    if i % 5 == 0 and i > 100:
        print(i)
    
```

### Câu 4: Định dạng đường thẳng

Phát sinh x là các điểm dữ liệu từ $[0, \pi]$. Bạn hãy vẽ đồ thị $y = sin(x)$ và $y = cos(x)$ như dưới đây.
Gợi ý: sử dụng hằng số pi và hàm sau:
- `np.pi` : số $\pi$
- `np.linspace()`
- `np.sin()`
- `np.cos()`

```python
# This cell contains output. Don't remove!
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Phát sinh x từ 0 đến 2 * pi
x = np.linspace(0, 2 * np.pi, 100)

# Tính toán giá trị y cho sin(x) và cos(x)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Tạo đồ thị
plt.figure(figsize=(8, 5))

# Vẽ đồ thị sin(x)
plt.plot(x, y_sin, label='sin(x)', color='blue', linestyle='-')

# Vẽ đồ thị cos(x)
plt.plot(x, y_cos, label='cos(x)', color='orange', linestyle='--')

# Thêm tiêu đề, nhãn và chú thích
plt.title('Đồ thị hàm sin(x) và cos(x)')
plt.xlabel('x')
plt.ylabel('y')
plt.legend(loc='lower left')

# Hiển thị lưới
plt.grid()

# Hiển thị đồ thị
plt.show()
```

### Câu 5: Vẽ các đồ thị con

```python
# This cell contains output. Don't remove!
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Phát sinh x từ 0 đến 2 * pi
x = np.linspace(0, 2 * np.pi, 100)

# Tính toán giá trị y cho sin(x) và cos(x)
y_sin = np.sin(x)
y_cos = np.cos(x)

# Tạo figure và các subplots
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8))

# Đồ thị sin(x)
ax1.plot(x, y_sin, label='sin(x)', color='blue')
ax1.set_title('Đồ thị hàm sin(x)')
ax1.set_xlabel('x')
ax1.set_ylabel('y')
ax1.legend(loc='lower left')
ax1.grid()

# Đồ thị cos(x)
ax2.plot(x, y_cos, label='cos(x)', color='orange', linestyle='--')
ax2.set_title('Đồ thị hàm cos(x)')
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.legend(loc='lower left')
ax2.grid()

# Hiển thị đồ thị
plt.tight_layout()
plt.show()
```

### Câu 8: Vẽ histogram

Sử dụng hàm `np.random.normal(size=10000)` để phát sinh 10000 số ngẫu nghiên thuộc phân phối chuẩn. Vẽ histogram các số ngẫu nhiên trên.

```python
# This cell contains output. Don't remove!
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Phát sinh 1000 số ngẫu nhiên có phân phối chuẩn với mean=0 và std=1

x = np.random.normal(0, 1, 1000)
# Vẽ histogram với số cột là 50
plt.hist(x, bins=100)

# Thêm các nhãn và tiêu đề
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')
plt.ylim(0, 350)
# Hiển thị biểu đồ
plt.show()
```
