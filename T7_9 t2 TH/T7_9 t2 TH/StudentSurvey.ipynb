```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('StudentSurvey.csv')
df.head()
```

```python


# 1. Bảng phân phối tần số số lượng sinh viên theo từng năm học
year_distribution = df["Year"].dropna().value_counts()
print("Bảng phân phối tần số:\n", year_distribution)

# 2a. Tỷ lệ sinh viên đeo khuyên
piercing_ratio = df[df["Piercings"] > 0].shape[0] / len(df)
print(f"Tỷ lệ sinh viên đeo khuyên: {piercing_ratio:.2%}")

# 2b. Tỷ lệ đeo khuyên giữa nam và nữ
male_piercing_ratio = df[(df["Gender"] == "M") & (df["Piercings"] > 0)].shape[0] / df[df["Gender"] == "M"].shape[0]
female_piercing_ratio = df[(df["Gender"] == "F") & (df["Piercings"] > 0)].shape[0] / df[df["Gender"] == "F"].shape[0]

print(f"Tỷ lệ sinh viên nam đeo khuyên: {male_piercing_ratio:.2%}")
print(f"Tỷ lệ sinh viên nữ đeo khuyên: {female_piercing_ratio:.2%}")

# 3. Vẽ histogram
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

sns.histplot(df["Pulse"].dropna(), bins=20, kde=True, ax=axes[0], color="blue")
axes[0].set_title("Histogram của nhịp tim")

sns.histplot(df["Exercise"].dropna(), bins=20, kde=True, ax=axes[1], color="green")
axes[1].set_title("Histogram của số giờ tập thể thao")

sns.histplot(df["Piercings"].dropna(), bins=20, kde=True, ax=axes[2], color="red")
axes[2].set_title("Histogram của số khuyên đeo")

plt.tight_layout()
plt.show()
```
