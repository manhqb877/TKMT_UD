**Bài 01:** Một bài thi trắc nghiệm gồm 10 câu hỏi, mỗi câu có 4 phương án trả lời trong đó chỉ có một phương án đúng. Bạn hãy: 

a. Một sinh viên không học bài làm bài bằng cách chọn ngẫu nhiên một phương án cho mỗi câu hỏi. Bạn hãy dùng hàm random để in ra số câu đúng của bạn sinh viên trong 10 lần kiểm tra.

b. Giả sử mỗi câu đúng được 4 điểm, mỗi câu sai bị trừ 2 điểm. Tính xác suất để sinh viên này được 4 điểm

c. Gọi X là số câu trả lời đúng. Tính E(X) và Var(X)

```python
# a, 
import random

def count_correct_answers():
    answers = [1, 0, 0, 0]  
    random.shuffle(answers) 

    return answers[0]  
# Số lần kiểm tra
num_tests = 10
correct_count = 0

# Thực hiện kiểm tra 10 lần
for _ in range(num_tests):
    correct_count += count_correct_answers()

print(correct_count) 
```

```python
import math

# Hàm tính xác suất
def probability_of_getting_4_points():
    # C(10, 4) * (1/4)^4 * (3/4)^6
    n = 10  # số câu hỏi
    k = 4   # số câu đúng cần thiết
    p_correct = 1 / 4  # xác suất trả lời đúng mỗi câu
    p_wrong = 3 / 4    # xác suất trả lời sai mỗi câu

    # Tính số cách chọn 4 câu đúng từ 10 câu
    combinations = math.comb(n, k)

    # Tính xác suất
    probability = combinations * (p_correct**k) * (p_wrong**(n - k))
    return probability

# Tính xác suất
probability = probability_of_getting_4_points()
print(f"Xác suất để sinh viên đạt được 4 điểm là: {probability:.6f}")
```

```python
# E(X)=n⋅p, p = 1/4
# Var(X)=n⋅p⋅(1−p)

E_X = 10*1/4
Var_x = (10 *1/4) *3/4
print({E_X})
print({Var_x})
```

**Bài 02:** Một trung tâm bưu điện nhận được 3 cuộc gọi điện thoại mỗi phút. Tính xác suất để trung tâm nhận được 1 cuộc gọi, 2 cuộc gọi, 3 cuộc gọi trong vòng một phút biết số cuộc gọi trong một phút có phân phối Poisson.

```python
import math

# Hàm tính xác suất phân phối Poisson
def poisson_probability(lam, k):
    return (lam**k * math.exp(-lam)) / math.factorial(k)

# Tần suất trung bình (lambda)
lambda_ = 3

P_1 = poisson_probability(lambda_, 1)
P_2 = poisson_probability(lambda_, 2)
P_3 = poisson_probability(lambda_, 3)

# In kết quả
print(f"Xác suất để nhận được 1 cuộc gọi: {P_1:.4f}")
print(f"Xác suất để nhận được 2 cuộc gọi: {P_2:.4f}")
print(f"Xác suất để nhận được 3 cuộc gọi: {P_3:.4f}")
```

**Bài 03:** Trọng lượng (đơn vị gam) của một loại trái cây có phân phối chuẩn với µ = 500 (gam) gam và $σ^2 = 16 (gam^2)$. Trái cây thu hoạch được phân loại theo trọng lượng như sau:

Loại 1: trên 505 gam

Loại 2: từ 495 – 505 gam

Loại 3: dười 495 gam

a. Hãy tính tỷ lệ của mỗi loại.
b. Bạn hãy mô phỏng lấy mẫu  20 trái cây trên. Bạn hãy vẽ histogram trọng lượng trái cây và tính giá trị trung bình và độ lệch chuẩn trái cây từ mẫu
c. Bạn hãy mô phỏng lấy 100 trái cây trên. Bạn hãy vẽ histogram trọng lượng trái cây và tính giá trị trung bình và độ lệch chuẩn trái cây từ mẫu
d. Từ hai cách lấy mẫu ở câu b, c bạn có nhận xét gì không?

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

# Các tham số của phân phối chuẩn
mu = 500    # Trung bình
sigma = 4   # Độ lệch chuẩn
sample_size_20 = 20
sample_size_100 = 100

# a. Tính tỷ lệ của mỗi loại
# Tỷ lệ Loại 1: P(X > 505)
P_X_greater_than_505 = 1 - norm.cdf(505, loc=mu, scale=sigma)

# Tỷ lệ Loại 2: P(495 <= X <= 505)
P_495_to_505 = norm.cdf(505, loc=mu, scale=sigma) - norm.cdf(495, loc=mu, scale=sigma)

# Tỷ lệ Loại 3: P(X < 495)
P_X_less_than_495 = norm.cdf(495, loc=mu, scale=sigma)

# In tỷ lệ các loại
print(f"Tỷ lệ Loại 1 (X > 505): {P_X_greater_than_505:.4f}")
print(f"Tỷ lệ Loại 2 (495 <= X <= 505): {P_495_to_505:.4f}")
print(f"Tỷ lệ Loại 3 (X < 495): {P_X_less_than_495:.4f}")

# b. Mô phỏng lấy mẫu 20 trái cây và vẽ histogram
sample_20 = np.random.normal(mu, sigma, sample_size_20)

# Tính giá trị trung bình và độ lệch chuẩn của mẫu 20
mean_20 = np.mean(sample_20)
std_20 = np.std(sample_20)

# Vẽ histogram mẫu 20
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.hist(sample_20, bins=8, edgecolor='black', alpha=0.7)
plt.title('Histogram - Mẫu 20 trái cây')
plt.xlabel('Trọng lượng (gam)')
plt.ylabel('Tần suất')

# c. Mô phỏng lấy mẫu 100 trái cây và vẽ histogram
sample_100 = np.random.normal(mu, sigma, sample_size_100)

# Tính giá trị trung bình và độ lệch chuẩn của mẫu 100
mean_100 = np.mean(sample_100)
std_100 = np.std(sample_100)

# Vẽ histogram mẫu 100
plt.subplot(1, 2, 2)
plt.hist(sample_100, bins=8, edgecolor='black', alpha=0.7)
plt.title('Histogram - Mẫu 100 trái cây')
plt.xlabel('Trọng lượng (gam)')
plt.ylabel('Tần suất')

plt.tight_layout()
plt.show()

# In giá trị trung bình và độ lệch chuẩn
print(f"Trung bình của mẫu 20 trái cây: {mean_20:.2f} gam")
print(f"Độ lệch chuẩn của mẫu 20 trái cây: {std_20:.2f} gam")

print(f"Trung bình của mẫu 100 trái cây: {mean_100:.2f} gam")
print(f"Độ lệch chuẩn của mẫu 100 trái cây: {std_100:.2f} gam")
```

**Bài 04:** Mô phỏng tung một con xúc sắc cân đối đồng chất 5000 lần. Dựa vào giá trị mô phỏng, bạn hãy tìm các giá trị xác suất dưới đây:

a. Xác suất để  số chấm xuất hiện là 4

b. Xác suất để số chấm xuất hiện lớn hơn hoặc bằng 4

c. Giả sử biết số chấm xuất hiện lớn hơn hoặc bằng 4. Hãy tìm xác suất để mặt 6 chấm xuất hiện.

```python
import numpy as np

# Mô phỏng tung xúc xắc 5000 lần
n_tung = 5000
result = np.random.randint(1, 7, size=n_tung)  # Các kết quả tung từ 1 đến 6

# a. Xác suất để số chấm xuất hiện là 4
count_4 = np.sum(result == 4)
prob_4 = count_4 / n_tung
print(f"a. Xác suất để số chấm xuất hiện là 4: {prob_4:.4f}")

# b. Xác suất để số chấm xuất hiện lớn hơn hoặc bằng 4
count_ge_4 = np.sum(result >= 4)
prob_ge_4 = count_ge_4 / n_tung
print(f"b. Xác suất để số chấm xuất hiện lớn hơn hoặc bằng 4: {prob_ge_4:.4f}")

# c. Xác suất để mặt 6 chấm xuất hiện, biết số chấm lớn hơn hoặc bằng 4
count_6_given_ge_4 = np.sum((result == 6) & (result >= 4))
prob_6_given_ge_4 = count_6_given_ge_4 / count_ge_4
print(f"c. Xác suất để mặt 6 chấm xuất hiện, biết số chấm lớn hơn hoặc bằng 4: {prob_6_given_ge_4:.4f}")
```
