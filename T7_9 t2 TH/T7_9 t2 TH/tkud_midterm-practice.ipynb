
### Câu 5

Phân bố tuổi tác trong mẫu: Hãy mô tả phân bố tuổi của người tham gia khảo sát. Tuổi trung bình, độ lệch chuẩn, và các phân vị 25%, 50%, 75% là bao nhiêu?



```python
import pandas as pd

# Giả sử bạn đã có một DataFrame df với cột 'age' chứa tuổi của người tham gia khảo sát
# Ví dụ về cách tạo một DataFrame mẫu (nếu cần):
# df = pd.DataFrame({'age': [23, 45, 56, 33, 25, 40, 52, 60, 37, 30, 29, 41]})

# Tính tuổi trung bình (mean)
mean_age = df['age'].mean()

# Tính độ lệch chuẩn (std)
std_dev_age = df['age'].std()

# Tính các phân vị 25%, 50%, 75%
q25 = df['age'].quantile(0.25)
q50 = df['age'].quantile(0.50)  # hoặc df['age'].median() cho phân vị 50%
q75 = df['age'].quantile(0.75)

# In kết quả
print(f"Tuổi trung bình: {mean_age:.2f}")
print(f"Độ lệch chuẩn: {std_dev_age:.2f}")
print(f"Phân vị 25%: {q25:.2f}")
print(f"Phân vị 50%: {q50:.2f}")
print(f"Phân vị 75%: {q75:.2f}")
```

### Câu 7

BMI trung bình theo giới tính: Tính chỉ số BMI trung bình cho nam và nữ trong mẫu. 



```python
import pandas as pd

# Giả sử bạn có một DataFrame df với các cột 'gender' (giới tính), 'height_cm' (chiều cao cm) và 'weight_kg' (cân nặng kg)
# Ví dụ dữ liệu:
df = pd.DataFrame({
    'gender': ['Nam', 'Nữ', 'Nam', 'Nữ', 'Nam'],
    'height_cm': [170, 160, 175, 150, 180],  # Chiều cao (cm)
    'weight_kg': [70, 55, 80, 45, 85]  # Cân nặng (kg)
})

# Tính BMI (Chiều cao cần chuyển từ cm sang m)
df['BMI'] = df['weight_kg'] / (df['height_cm'] / 100) ** 2

# Tính BMI trung bình theo giới tính
average_bmi_by_gender = df.groupby('gender')['BMI'].mean()

# In kết quả
print(average_bmi_by_gender)
```
