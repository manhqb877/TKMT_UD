4. Viết chương trình yêu cầu người dùng nhập họ tên. Hiển thị tên người dùng sau khi đã chuẩn hóa. Biết rằng tên đã chuẩn hóa là:
- Ký tự đầu viết hoa
- Không có khoảng trắng thừa
- Không có ký tự đặc biệt

````python
import re

def chuan_hoa_ten(ten):
    # Loại bỏ ký tự đặc biệt
    ten = re.sub(r'[^a-zA-Z\s]', '', ten)
    # Loại bỏ khoảng trắng thừa
    ten = re.sub(r'\s+', ' ', ten).strip()
    # Viết hoa ký tự đầu của mỗi từ
    ten = ' '.join(word.capitalize() for word in ten.split())
    return ten

# Yêu cầu người dùng nhập họ tên
s = input('Nhập vào họ tên: ')

# Chuẩn hóa tên
ten_chuan_hoa = chuan_hoa_ten(s)

# Hiển thị kết quả
print(f'Tên đã chuẩn hóa: {ten_chuan_hoa}')
````

5. Viết chương trình nhập vào một số nguyên dương (n>0). Cho biết n có phải là số đối xứng hay không? Biết số đối xứng là số viết xuôi hay viết ngược đều như nhau

````python
n = int(input('Nhập vào số đối xứng: '))
print("Số đã nhập:", n)

s = str(n)
s_reversed = s[::-1]  # Đảo ngược chuỗi
print("Chuỗi đảo ngược:", s_reversed)

if s == s_reversed:
    print(f"{n} là số đối xứng.")
else:
    print(f"{n} không phải là số đối xứng.")
````

8. Viết chương trình chấp nhận một chuỗi số, phân tách bằng dấu phẩy từ giao diện điều khiển, tạo ra một List và một tuple chứa mọi số.

    Ví dụ:

    ```
    Đầu vào được cung cấp là '34,67,55,33,12,98' thì đầu ra là:

    [34, 67, 55, 33, 12, 98]
    (34, 67, 55, 33, 12, 98)
    ```


````python
# Nhập chuỗi số, phân tách bằng dấu phẩy
input_data = input("Nhập chuỗi số, phân tách bằng dấu phẩy: ")

# Tạo List từ chuỗi đầu vào
num_list = [int(i) for i in input_data.split(",")]

# Tạo Tuple từ List
num_tuple = tuple(num_list)

# Hiển thị kết quả
print("Danh sách List:", num_list)
print("Danh sách Tuple:", num_tuple)
````

9. Viết hàm nhận vào một danh sách các số nguyên. Hãy viết hàm trả về một danh sách gồm các số không trùng nhau trong danh sách ban đầu

````python
def unique_numbers(numbers):
    """
    Trả về danh sách các số không trùng lặp từ danh sách đầu vào.

    :param numbers: Danh sách các số nguyên
    :return: Danh sách các số không trùng lặp
    """
    return list(set(numbers))

# Ví dụ sử dụng
input_list = [1, 2, 2, 3, 4, 4, 5]
result = unique_numbers(input_list)
print("Danh sách không trùng lặp:", result)
````

10. Viết hàm có tham số đầu vào là một list L có các phần tử là chuỗi. Hãy tạo ra một dictionary D mã hóa, với mỗi một phần tử trong L được mã hóa thành một con số (theo thứ tự từ 0 tăng dần lên 1 đơn vị). Sau đó trả về list đã được mã hóa

    Ví dụ:

    ```
    Cho L = ["đen","vàng","xanh","vàng","xanh","đỏ","hồng"]

    Xây dựng dictionary mã hóa:
    D = {'đen':0,'vàng':1,'xanh':2,'đỏ':3,'hồng':4}

    Trả về List mã hóa:
    L_mahoa = [0, 1, 2, 1, 2, 3, 4]
    ```

````python
def encode_list(L):
    """
    Mã hóa danh sách các chuỗi thành danh sách các số dựa trên dictionary mã hóa.

    :param L: Danh sách các chuỗi
    :return: Dictionary mã hóa và danh sách đã được mã hóa
    """
    # Tạo dictionary mã hóa
    D = {value: index for index, value in enumerate(sorted(set(L)))}

    # Mã hóa danh sách
    L_encoded = [D[item] for item in L]

    return D, L_encoded

# Ví dụ sử dụng
L = ["đen", "vàng", "xanh", "vàng", "xanh", "đỏ", "hồng"]
D, L_mahoa = encode_list(L)

# Hiển thị kết quả
print("Dictionary mã hóa:", D)
print("Danh sách đã mã hóa:", L_mahoa)
````
